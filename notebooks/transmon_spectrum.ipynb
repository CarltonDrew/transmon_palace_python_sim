```python
import numpy as np
import matplotlib.pyplot as plt
import scqubits as transmons

# Initialize a standard Transmon qubit circuit
qubit = transmons.Transmon(
    EJ=57.2,
    EC=1.0,
    ng=0.0,
    ncut=30
)

# 1. Define the range of offset charges to sweep over (from -2 to 2)
ng_list = np.linspace(-2, 2, 220)

# 2. Call the correct generalized method to sweep and plot the energy spectrum
qubit.plot_evals_vs_paramvals(
    param_name='ng', 
    param_vals=ng_list, 
    evals_count=4, 
    subtract_ground=False
)

plt.show()
```
